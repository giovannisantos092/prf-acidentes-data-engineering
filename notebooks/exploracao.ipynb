## Conversão da data

Converte a coluna de data para o formato de data do Pandas.

```python
df["data_inversa"] = pd.to_datetime(df["data_inversa"])
```

## Verificação do tipo da data

Confirma se a coluna foi convertida corretamente.

```python
df["data_inversa"].dtype
```

## Extração do ano

Cria uma coluna com o ano do acidente a partir da data.

```python
df["ano"] = df["data_inversa"].dt.year
```

## Extração do mês

Cria uma coluna com o mês do acidente.

```python
df["mes"] = df["data_inversa"].dt.month
```

## Acidentes por mês

Mostra a quantidade de acidentes registrados em cada mês.

```python
df["mes"].value_counts().sort_index()
```
