## Padronização dos tipos de dados

Converte datas e campos numéricos para tipos adequados antes da união das bases.

```python
def limpar_dados(df):
    df = df.copy()

    # Tratamento dos valores nulos
    df["classificacao_acidente"] = (
        df["classificacao_acidente"]
        .fillna("Não informado")
    )

    colunas_administrativas = [
        "regional",
        "delegacia",
        "uop"
    ]

    df[colunas_administrativas] = (
        df[colunas_administrativas]
        .fillna("Não informado")
    )

    # Padronização do ID
    df["id"] = df["id"].astype("Int64")

    # Conversão da data
    df["data_inversa"] = pd.to_datetime(
        df["data_inversa"],
        errors="coerce"
    )

    # Conversão do KM
    df["km"] = (
        df["km"]
        .astype(str)
        .str.replace(",", ".", regex=False)
    )

    df["km"] = pd.to_numeric(
        df["km"],
        errors="coerce"
    )

    # Conversão de latitude e longitude
    for coluna in ["latitude", "longitude"]:
        df[coluna] = (
            df[coluna]
            .astype(str)
            .str.replace(",", ".", regex=False)
        )

        df[coluna] = pd.to_numeric(
            df[coluna],
            errors="coerce"
        )

    return df
```

## Validação das conversões em todas as bases

Aplica a função de limpeza em todos os anos e verifica se as conversões geraram valores nulos.

```python
colunas_convertidas = [
    "id",
    "data_inversa",
    "km",
    "latitude",
    "longitude"
]

for ano, caminho in arquivos.items():
    df_ano = pd.read_csv(
        caminho,
        sep=";",
        encoding="latin1"
    )

    df_ano = limpar_dados(df_ano)

    print(f"\nAno: {ano}")
    print(df_ano[colunas_convertidas].isnull().sum())
```

## Registros com falha nas conversões

Conta quantos registros possuem valores ausentes nos campos numéricos e geográficos após a conversão.

```python
for ano, caminho in arquivos.items():
    df_ano = pd.read_csv(
        caminho,
        sep=";",
        encoding="latin1"
    )

    df_ano = limpar_dados(df_ano)

    problemas = df_ano[
        df_ano[
            ["km", "latitude", "longitude"]
        ].isnull().any(axis=1)
    ]

    print(
        f"{ano} -> "
        f"{len(problemas)} registros com problema de conversão"
    )
```

## União das bases

Aplica a limpeza em todos os anos e reúne as bases de 2022 a 2026 em um único DataFrame.

```python
bases_processadas = []

for ano, caminho in arquivos.items():
    df_ano = pd.read_csv(
        caminho,
        sep=";",
        encoding="latin1"
    )

    df_ano = limpar_dados(df_ano)

    bases_processadas.append(df_ano)

df_prf = pd.concat(
    bases_processadas,
    ignore_index=True
)
```
